```python
# ============================================================
# AI DECISION SUPPORT ASSISTANT
# CELL 2 — DECISION ANALYSIS ENGINE
# ============================================================

import re


# ------------------------------------------------------------
# DECISION ANALYZER
# ------------------------------------------------------------

def analyze_decision(situation, option_a, option_b):

    if not situation.strip():
        return "No situation provided."

    if not option_a.strip() or not option_b.strip():
        return "Please provide both options."

    # --------------------------------------------------------
    # Extract important words
    # --------------------------------------------------------

    positive_words = [
        "benefit", "advantage", "improve", "growth",
        "save", "learn", "opportunity", "flexible",
        "experience", "valuable", "better", "increase"
    ]

    risk_words = [
        "expensive", "cost", "risk", "difficult",
        "stress", "delay", "problem", "limited",
        "uncertain", "loss", "far", "time"
    ]

    def evaluate_option(option):

        text = option.lower()

        positive_score = sum(
            1 for word in positive_words
            if re.search(r"\b" + re.escape(word) + r"\b", text)
        )

        risk_score = sum(
            1 for word in risk_words
            if re.search(r"\b" + re.escape(word) + r"\b", text)
        )

        total = 50 + (positive_score * 8) - (risk_score * 6)

        return max(0, min(100, total)), positive_score, risk_score

    score_a, benefits_a, risks_a = evaluate_option(option_a)
    score_b, benefits_b, risks_b = evaluate_option(option_b)

    # --------------------------------------------------------
    # Recommendation
    # --------------------------------------------------------

    if score_a > score_b:
        recommendation = "OPTION A"
        difference = score_a - score_b

    elif score_b > score_a:
        recommendation = "OPTION B"
        difference = score_b - score_a

    else:
        recommendation = "BOTH OPTIONS"
        difference = 0

    # --------------------------------------------------------
    # Confidence
    # --------------------------------------------------------

    if difference >= 25:
        confidence = "High"

    elif difference >= 12:
        confidence = "Moderate"

    else:
        confidence = "Low"

    # --------------------------------------------------------
    # Generate analysis
    # --------------------------------------------------------

    report = f"""
DECISION ANALYSIS
────────────────────────────────────────

Situation:
{situation}

OPTION A
────────────────────────────────────────
{option_a}

Estimated Score: {score_a}/100
Positive Indicators: {benefits_a}
Risk Indicators: {risks_a}


OPTION B
────────────────────────────────────────
{option_b}

Estimated Score: {score_b}/100
Positive Indicators: {benefits_b}
Risk Indicators: {risks_b}


RECOMMENDATION
────────────────────────────────────────
{recommendation}

Confidence Level: {confidence}

The recommendation is based on the information
provided in the two options.


IMPORTANT
────────────────────────────────────────
This is a decision-support tool, not an automatic
decision maker. Consider your personal priorities,
constraints, risks and real-world circumstances before
making an important decision.
"""

    return report.strip()


# ------------------------------------------------------------
# TEST
# ------------------------------------------------------------

situation = """
I want to choose between two internship opportunities
during my semester break.
"""

option_a = """
Option A offers practical AI experience, flexible timing,
good learning opportunities and valuable project experience.
"""

option_b = """
Option B has a higher cost, requires long travel and may
cause time and stress issues, but provides good experience.
"""

result = analyze_decision(
    situation,
    option_a,
    option_b
)

print(result)
```

```python
# ============================================================
# AI DECISION SUPPORT ASSISTANT
# CELL 3 — SMART INSIGHT ENGINE
# ============================================================

import re


def generate_ai_insights(situation, option_a, option_b):

    if not situation or not situation.strip():
        return "Please enter your situation."

    if not option_a or not option_a.strip():
        return "Please enter Option A."

    if not option_b or not option_b.strip():
        return "Please enter Option B."

    # --------------------------------------------------------
    # Keyword categories
    # --------------------------------------------------------

    career_words = [
        "career", "job", "internship", "experience",
        "skill", "learning", "growth", "future"
    ]

    financial_words = [
        "cost", "money", "salary", "stipend",
        "price", "expense", "save", "budget"
    ]

    time_words = [
        "time", "schedule", "travel", "distance",
        "flexible", "duration", "hours"
    ]

    risk_words = [
        "risk", "difficult", "stress", "problem",
        "uncertain", "expensive", "delay", "loss"
    ]

    # --------------------------------------------------------
    # Detect priorities from situation
    # --------------------------------------------------------

    situation_lower = situation.lower()

    detected_priorities = []

    if any(word in situation_lower for word in career_words):
        detected_priorities.append("Career & Learning")

    if any(word in situation_lower for word in financial_words):
        detected_priorities.append("Financial Impact")

    if any(word in situation_lower for word in time_words):
        detected_priorities.append("Time & Convenience")

    if not detected_priorities:
        detected_priorities.append("Overall Benefit & Risk")

    # --------------------------------------------------------
    # Analyze each option
    # --------------------------------------------------------

    def analyze_option(text):

        text_lower = text.lower()

        benefits = []
        risks = []

        for word in career_words:
            if re.search(r"\b" + re.escape(word) + r"\b", text_lower):
                benefits.append(word)

        for word in financial_words:
            if re.search(r"\b" + re.escape(word) + r"\b", text_lower):
                benefits.append(word)

        for word in time_words:
            if re.search(r"\b" + re.escape(word) + r"\b", text_lower):
                benefits.append(word)

        for word in risk_words:
            if re.search(r"\b" + re.escape(word) + r"\b", text_lower):
                risks.append(word)

        return benefits, risks

    benefits_a, risks_a = analyze_option(option_a)
    benefits_b, risks_b = analyze_option(option_b)

    # --------------------------------------------------------
    # Generate practical insights
    # --------------------------------------------------------

    insights = []

    if len(benefits_a) > len(benefits_b):
        insights.append(
            "Option A contains more positive indicators "
            "based on the information provided."
        )

    elif len(benefits_b) > len(benefits_a):
        insights.append(
            "Option B contains more positive indicators "
            "based on the information provided."
        )

    else:
        insights.append(
            "Both options contain a similar number of "
            "positive indicators."
        )

    if len(risks_a) > len(risks_b):
        insights.append(
            "Option A shows comparatively more potential "
            "risk indicators."
        )

    elif len(risks_b) > len(risks_a):
        insights.append(
            "Option B shows comparatively more potential "
            "risk indicators."
        )

    else:
        insights.append(
            "Both options show a similar level of risk indicators."
        )

    # --------------------------------------------------------
    # Recommended evaluation method
    # --------------------------------------------------------

    if "Career & Learning" in detected_priorities:
        next_step = (
            "Compare learning opportunities, practical experience, "
            "mentorship and long-term career value."
        )

    elif "Financial Impact" in detected_priorities:
        next_step = (
            "Compare total cost, expected financial benefit and "
            "whether the decision fits your budget."
        )

    elif "Time & Convenience" in detected_priorities:
        next_step = (
            "Compare travel, schedule, time commitment and "
            "overall convenience."
        )

    else:
        next_step = (
            "List your top three priorities and compare both "
            "options against each priority."
        )

    # --------------------------------------------------------
    # Final report
    # --------------------------------------------------------

    report = f"""
🧠 SMART DECISION INSIGHTS
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

🎯 MAIN PRIORITIES
{", ".join(detected_priorities)}

🔹 OPTION A INSIGHT
Positive indicators: {len(benefits_a)}
Potential risk indicators: {len(risks_a)}

🔹 OPTION B INSIGHT
Positive indicators: {len(benefits_b)}
Potential risk indicators: {len(risks_b)}

💡 KEY INSIGHTS
• {insights[0]}
• {insights[1]}

🚀 RECOMMENDED NEXT STEP
{next_step}

📌 DECISION TIP
Do not choose an option only because it has more
advantages. Consider your personal priorities,
constraints, risks and long-term goals before making
an important decision.

⚠️ This system provides decision-support insights only.
It does not make decisions on behalf of the user.
"""

    return report.strip()


# ============================================================
# TEST
# ============================================================

test_situation = """
I want to choose between two internship opportunities
for my career development.
"""

test_option_a = """
This internship provides AI experience, practical projects,
good mentorship and flexible working hours.
"""

test_option_b = """
This internship offers a higher stipend but requires
long travel and has limited technical learning.
"""

result = generate_ai_insights(
    test_situation,
    test_option_a,
    test_option_b
)

print(result)
```

```python
# ============================================================
# AI DECISION SUPPORT ASSISTANT
# CELL 4 — PROFESSIONAL INTERACTIVE DASHBOARD
# ============================================================

import gradio as gr


# ------------------------------------------------------------
# COMPLETE ANALYSIS
# ------------------------------------------------------------

def run_decision_analysis(situation, option_a, option_b):

    if not situation or not situation.strip():
        return (
            0,
            0,
            "Waiting for input",
            "Please describe your situation."
        )

    if not option_a or not option_a.strip():
        return (
            0,
            0,
            "Option A missing",
            "Please enter details for Option A."
        )

    if not option_b or not option_b.strip():
        return (
            0,
            0,
            "Option B missing",
            "Please enter details for Option B."
        )

    # Existing decision engine
    report = analyze_decision(
        situation,
        option_a,
        option_b
    )

    # Extract scores using regex
    scores = re.findall(
        r"Estimated Score:\s*(\d+)/100",
        report
    )

    if len(scores) >= 2:
        score_a = int(scores[0])
        score_b = int(scores[1])
    else:
        score_a = 0
        score_b = 0

    # Smart insights
    insights = generate_ai_insights(
        situation,
        option_a,
        option_b
    )

    # Recommendation
    if score_a > score_b:
        recommendation = "OPTION A"
    elif score_b > score_a:
        recommendation = "OPTION B"
    else:
        recommendation = "BOTH OPTIONS"

    final_report = report + "\n\n" + insights

    return (
        score_a,
        score_b,
        recommendation,
        final_report
    )


# ------------------------------------------------------------
# SAMPLE DATA
# ------------------------------------------------------------

def load_sample():

    situation = """
I have two internship opportunities and I want to
choose the one that will help my career development.
"""

    option_a = """
AI internship with practical projects, mentorship,
hands-on experience, flexible timing and good learning
opportunities.
"""

    option_b = """
Software internship with a higher stipend but longer
travel, limited technical work and a fixed schedule.
"""

    return situation, option_a, option_b


# ------------------------------------------------------------
# CLEAR
# ------------------------------------------------------------

def clear_all():

    return (
        "",
        "",
        "",
        0,
        0,
        "Waiting for analysis",
        ""
    )


# ============================================================
# CUSTOM CSS
# ============================================================

css = """

body {
    background: #f4efe8 !important;
}

.gradio-container {
    max-width: 1250px !important;
    margin: auto !important;
    font-family: Arial, sans-serif !important;
}

/* HERO */

.hero {
    background: #4a2028;
    color: #fff8ef;
    padding: 38px;
    border-radius: 24px;
    margin-bottom: 22px;
    box-shadow: 0 14px 35px rgba(74,32,40,0.18);
}

.hero h1 {
    font-size: 40px;
    margin: 0 0 8px 0;
}

.hero p {
    color: #eadbd0;
    font-size: 16px;
    margin: 6px 0;
}

/* PANELS */

.panel {
    background: #fffaf4;
    border: 1px solid #e2d6ca;
    border-radius: 20px;
    padding: 24px;
    box-shadow: 0 8px 25px rgba(70,45,35,0.08);
}

/* SCORE CARDS */

.score-card {
    background: #f8eee5;
    border: 1px solid #dfc8b7;
    border-radius: 18px;
}

/* RECOMMENDATION */

.recommendation {
    background: #4a2028;
    border-radius: 18px;
    padding: 18px;
}

/* BUTTONS */

.main-btn button {
    background: #8b3f4b !important;
    color: white !important;
    border: none !important;
    border-radius: 11px !important;
    font-weight: 700 !important;
}

.secondary-btn button {
    background: #eadbd0 !important;
    color: #4a2028 !important;
    border: none !important;
    border-radius: 11px !important;
    font-weight: 600 !important;
}

/* INFO */

.info {
    background: #f1e7dc;
    border-left: 5px solid #8b3f4b;
    padding: 18px;
    border-radius: 12px;
    color: #513c3c;
}

/* FOOTER */

.footer {
    text-align: center;
    color: #806f68;
    padding: 22px;
    font-size: 13px;
}

"""


# ============================================================
# GRADIO APPLICATION
# ============================================================

with gr.Blocks(
    title="DecisionLens AI",
    css=css
) as demo:

    # --------------------------------------------------------
    # HEADER
    # --------------------------------------------------------

    gr.HTML("""
    <div class="hero">

        <h1>🧭 DecisionLens AI</h1>

        <p>
            <b>Smart Decision Support & Insight Platform</b>
        </p>

        <p>
            Compare alternatives, identify important factors,
            evaluate risks and receive structured decision insights.
        </p>

    </div>
    """)

    # --------------------------------------------------------
    # INPUT AREA
    # --------------------------------------------------------

    with gr.Group(elem_classes="panel"):

        gr.Markdown("## 📝 Define Your Decision")

        situation = gr.Textbox(
            label="What decision are you trying to make?",
            placeholder=(
                "Example: I have two internship opportunities "
                "and want to choose the better one."
            ),
            lines=4
        )

        with gr.Row():

            option_a = gr.Textbox(
                label="🔹 Option A",
                placeholder=(
                    "Describe the first option, "
                    "including benefits or concerns."
                ),
                lines=7
            )

            option_b = gr.Textbox(
                label="🔸 Option B",
                placeholder=(
                    "Describe the second option, "
                    "including benefits or concerns."
                ),
                lines=7
            )

        with gr.Row():

            analyze_btn = gr.Button(
                "⚡ Analyze Decision",
                elem_classes="main-btn"
            )

            sample_btn = gr.Button(
                "📄 Try Sample",
                elem_classes="secondary-btn"
            )

            clear_btn = gr.Button(
                "↻ Reset",
                elem_classes="secondary-btn"
            )

    # --------------------------------------------------------
    # SCORE AREA
    # --------------------------------------------------------

    gr.Markdown("## 📊 Comparative Analysis")

    with gr.Row():

        score_a = gr.Number(
            label="Option A Score / 100",
            value=0,
            interactive=False,
            elem_classes="score-card"
        )

        score_b = gr.Number(
            label="Option B Score / 100",
            value=0,
            interactive=False,
            elem_classes="score-card"
        )

        recommendation = gr.Textbox(
            label="🏆 Decision Insight",
            value="Waiting for analysis",
            interactive=False,
            elem_classes="recommendation"
        )

    # --------------------------------------------------------
    # REPORT
    # --------------------------------------------------------

    with gr.Group(elem_classes="panel"):

        gr.Markdown("## 🔎 Detailed Decision Report")

        report_output = gr.Textbox(
            label="AI Decision Analysis",
            lines=18,
            interactive=False
        )

    # --------------------------------------------------------
    # INFORMATION
    # --------------------------------------------------------

    gr.HTML("""
    <div class="info">

        <b>💡 DecisionLens methodology</b><br><br>

        The system examines the information you provide,
        identifies positive and risk indicators, compares
        alternatives and generates structured insights.

        <br><br>

        It is designed to <b>support</b> human decision-making,
        not replace personal judgment.

    </div>
    """)

    # --------------------------------------------------------
    # FOOTER
    # --------------------------------------------------------

    gr.HTML("""
    <div class="footer">
        DecisionLens AI • Intelligent Decision Support Platform
        <br>
        Educational AI demonstration project
    </div>
    """)

    # ========================================================
    # EVENTS
    # ========================================================

    analyze_btn.click(
        fn=run_decision_analysis,
        inputs=[
            situation,
            option_a,
            option_b
        ],
        outputs=[
            score_a,
            score_b,
            recommendation,
            report_output
        ]
    )

    sample_btn.click(
        fn=load_sample,
        outputs=[
            situation,
            option_a,
            option_b
        ]
    )

    clear_btn.click(
        fn=clear_all,
        outputs=[
            situation,
            option_a,
            option_b,
            score_a,
            score_b,
            recommendation,
            report_output
        ]
    )


# ============================================================
# LAUNCH
# ============================================================

demo.launch(
    share=True,
    debug=True
)
```
